```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd
from PIL import Image
```

```python
REPO = Path("..").resolve()
DATA = REPO / "data"
ANN = DATA / "annotations" / "task2" / "val"
assert ANN.exists(), (
    f"Missing {ANN}. Run from repo root:\n"
    "  uv run python scripts/download-dataset.py"
)
print("DATA =", DATA)
print("ANN  =", ANN)
print("contents:", sorted(p.name for p in ANN.iterdir()))
```

```python
gt_path = ANN / "gt.csv"

header = gt_path.read_text(encoding="utf-8").splitlines()[0]
cols = [c.strip() for c in header.lstrip("# ").split(",")]
ATTRS = cols[1:]  # remove col "image"

gt = pd.read_csv(gt_path, names=cols, comment="#")
print(f"{len(ATTRS)} attrs:", ATTRS[:5], "...")
print("gt.shape:", gt.shape)  # ~33407 x 41

gt.head(3)

```

```python
q_path = ANN / "queries.csv"
ids_path = ANN / "ids.csv"

q_header = q_path.read_text(encoding="utf-8").splitlines()[0]
q_cols = [c.strip() for c in q_header.lstrip("# ").split(",")]
queries = pd.read_csv(q_path, names=q_cols, comment="#")

ids = pd.read_csv(ids_path, names=["image", "query_id"], header=None)

print("queries:", queries.shape, "— mỗi hàng = 1 query 40-bit")
print("ids:    ", ids.shape, "— mỗi ảnh map → query_id")
print("n_queries (unique ids):", ids["query_id"].nunique())
ids.head(3)
```

```python
row = gt.iloc[0]
img_path = DATA / row["image"]

print("image:", row["image"])
print("exists:", img_path.exists())
print("active attrs:", [a for a in ATTRS if int(row[a]) == 1])

if img_path.exists():
    plt.figure(figsize=(3, 6))
    plt.imshow(Image.open(img_path))
    plt.axis("off")
    plt.show()
```
